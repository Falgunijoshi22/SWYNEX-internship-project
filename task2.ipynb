```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_clean = pd.read_csv("cleaned_cafe_sales.csv")

df_clean.head()
```

```python
print("Dataset shape:", df_clean.shape)

df_clean.info()

df_clean.describe()

df_clean.isnull().sum()
```

```python
df_clean["Transaction Date"] = pd.to_datetime(
    df_clean["Transaction Date"],
    errors="coerce"
)
```

```python
df_clean["Transaction Date"].dtype
```

```python
df_clean.info()
```

```python
df_clean["Year"] = df_clean["Transaction Date"].dt.year
```

```python
df_clean.head()
```

```python
df_clean.info()
```

```python
df_clean["Year"] = pd.to_datetime(
    df_clean["Year"],
    errors="coerce"
)
```

```python
df_clean.info()
```

```python
df_clean["Month"] = df_clean["Transaction Date"].dt.month_name()
```

```python
df_clean.info()
```

```python
df_clean.head(50)
```

```python
df_clean = df_clean.drop(columns=["Year"])
```

```python
df_clean.head(50)
```

```python
print("Total transactions:", df_clean["Transaction ID"].nunique())
print("Total recorded sales:", df_clean["Total Spent"].sum())
print("Average transaction value:", df_clean["Total Spent"].mean())
print("Median transaction value:", df_clean["Total Spent"].median())
print("Total quantity sold:", df_clean["Quantity"].sum())
print("Unique items:", df_clean["Item"].nunique())
```

```python
item_quantity = (
    df_clean.groupby("Item")["Quantity"]
    .sum()
    .dropna()
    .sort_values(ascending=False)
)

item_quantity.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Total Quantity Sold by Item")
plt.xlabel("Item")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
item_revenue = (
    df_clean.groupby("Item")["Total Spent"]
    .sum(min_count=1)
    .dropna()
    .sort_values(ascending=False)
)

item_revenue.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Total Revenue by Item")
plt.xlabel("Item")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
monthly_revenue = (
    df_clean.dropna(subset=["Transaction Date", "Total Spent"])
    .set_index("Transaction Date")["Total Spent"]
    .resample("MS")
    .sum(min_count=1)
)

monthly_revenue.plot(
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
